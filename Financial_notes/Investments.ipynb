# Investing in Philippines 101

### What to consider when choosing a PSE investment platform

- Regulation / legitimacy (SEC‑registered broker)

- Minimum deposit or initial investment

- Commission & fees (trade, sales tax, PSE/SCCP fees, etc.)

- Platform usability (web, mobile app, charting tools, research, conditional orders)

- Access to other products (ETFs, foreign stocks, mutual funds or bonds)

- Integration with banks (easier deposits/withdrawals)

- Customer support
