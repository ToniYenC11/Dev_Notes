## About the Data

It contains information on each student's study habits, motivation level, family history, attendance at school, and availability of learning materials. Lifestyle characteristics including physical activity, sleep duration, and involvement in extracurricular activities are also recorded in the dataset. Combining these factors gives the file a thorough understanding of how many social, educational, and personal factors affect the students' final exam results.

```python
import pandas as pd
import numpy as np

def explore(df):
    """Descriptive and exploratory analysis

    Args:
        df (pd.DataFrame): DataFrame to analyze
    """
    print("Head of dataset: ")
    display(df.head(5))
    
    print("Dataset information")
    display(df.info())
    
    print("Describe")
    display(df.describe())

def check_unqiue(df):
    """Check unqiue values and count null values for each columns

    Args:
        df (pd.DataFrame): DataFrame to analyze
    """
    null_cols = []
    for col in df.columns:
        print(f"Column {col}:")
        null_count = df[col].isna().sum()
        if null_count > 0:
            print(f"Null count: {null_count}\n")
            null_cols.append(col)
        else:
            print("\n")
            
        if df[col].dtype == "object":
            print(f"Unqiue values: {df[col].unique()} \n")
        else:
            print("\n")
    
    print(f"Null columns: {null_cols}")
```

```python
data_path = "StudentPerformanceFactors.csv"
data = pd.read_csv(data_path)

explore(data)
check_unqiue(data)
```
