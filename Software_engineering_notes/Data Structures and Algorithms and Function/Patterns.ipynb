# Pattern Recognition in Leetcode

## Time Complexity vs Input Size Guide

| Input Size (n) | Typical Upper Bound Algorithm                                |
| -------------- | ------------------------------------------------------------ |
| **n ≤ 10**     | (O(n!)), (O(2^n)) → brute force, backtracking                |
| **n ≤ 20–25**  | (O(2^n)) or (O(n·2^n)) → subsets, bitmask DP                 |
| **n ≤ 100**    | (O(n^3)) → Floyd-Warshall, 3-nested loops DP                 |
| **n ≤ 1,000**  | (O(n^2)) → DP, two-pointers, prefix sums                     |
| **n ≤ 10⁵**    | (O(n\log n)) → sorting, binary search, heap                  |
| **n ≤ 10⁶+**   | (O(n)) → hash maps, sliding window, prefix sums              |
| **n ≤ 10⁸+**  | (O(1)), (O(\log n)) → math formulas, binary search on answer |


## Constraint Patterns by Data Structure

| Data Type / Constraint        | Likely Pattern / Algorithm           |
| ----------------------------- | ------------------------------------ |
| **Sorted array**              | Binary search, two pointers          |
| **Distinct integers**         | No duplicates → direct mapping       |
| **Linked list**               | Slow/fast pointers, reversal         |
| **Binary tree**               | DFS, BFS, recursion                  |
| **Graph (edges ≤ 10⁴–10⁵)**   | BFS/DFS, Dijkstra, Union-Find        |
| **Matrix ≤ 100×100**          | DP, DFS/BFS                          |
| **Matrix ≤ 1000×1000**        | Prefix sums or optimized BFS         |
| **Modulo constraint (10⁹+7)** | Dynamic programming or combinatorics |
| **Subarray / substring**      | Sliding window, prefix sum           |
| **Interval merging**          | Sort + greedy                        |
| **K elements / K-th**         | Heap or binary search                |


## Subtle Constraints

| Constraint or Keyword      | Likely Insight                        |
| -------------------------- | ------------------------------------- |
| “at most once”, “distinct” | Use a set or frequency map            |
| “sorted”                   | Binary search or merge-like logic     |
| “continuous”               | Prefix sums or sliding window         |
| “adjacent”                 | DP relation or graph traversal        |
| “large range of values”    | Use coordinate compression            |
| “streaming input”          | Use heap or two heaps (median finder) |
| “mod 10⁹+7”                | Overflow-safe arithmetic or DP        |
| “up to 10⁴ test cases”     | Each test must be O(1) or O(log n)    |
