# Sorting ALgorithms

## Insertion Sort

```python
def insert_sort(arr,desc=False):
    for j in range(1, len(arr)):
        key = arr[j]
        i = j - 1
        
        compare = (lambda a, b: a < b) if desc else (lambda a, b: a > b)
        
        while i >= 0 and compare(arr[i],key): # desc
            arr[i+1] = arr[i]
            i -= 1
        arr[i+1] = key

#* Usage
my_list = [31, 41, 59, 26, 41, 58]
insert_sort(my_list,desc=True)
print(my_list)  
```

## Selection Sort

Pseudocode:

- Let first element as `min`

- Iterate through rest of list: `elem`:
    - If `elem` \< `min`: `arr[0]` = `elem`

# Searching Algorithms

## Linear Search

```python
def linear_search(arr,val):
    for i in range(len(arr)):
        if arr[i] == val:
            return i 
arr = [2,3,4,53,453,54]
val = 453
print(linear_search(arr,val))
```
