### Transactions in Databases

- A **transaction** is a collection of changes made to a database that must be performed as a unit.
- A transaction is not successful unless each of its operations is successful.
- At the database level, either all the database changes related to the transaction are performed, or no change is made to the database at all. It’s an all or nothing modification.

#### Status of a Transaction

- **Active** - Initial state when the transaction is being run.
- **Partially commited** - When it is completing its final operation.
- **Failed** - Any checks made by the database recovery system fail.
- **Aborted** - Database rolls back to original state before running a failed transaction.
- **Commited state** - When all operations within a transaction have been sucessfuly performed.

#### Transaction Use Cases

- Run a set of operations so that the database never contains the result of partial operations.
    - If one operation fails, the database is restored to its original state.
    - If no errors occur, the full set of statements changes the database.
- Provide isolation between programs that access a database simultaneously.

#### Properties of Transactions

ACID

- **Atomicity** - Ensures changes are successfully completed all at once or not at all.
- **Consistency** - Ensures any changes will not violate the integrity of the database, including any constraints.
- **Isolation** - Transactions are isolated so that they do not interfere with the other transactions.
- **Durability** - Ensures that change is permanent when a transaction is committed.

### Conditional Search

- You can use `UPPER` or `LOWER` with the like operator for exact string search that is not case-sensitive. To successfuly query "Cape Town" and "Fartown":

    ```SQL
    select name, district, population
    from city where
    lower(name) like '%town%';
    ```

- Using regular expressions with wildcards such as *%* and *_* for flexible search conditions.

- Regular expressions are available in all Amazon Relational Database Service (Amazon RDS) databases. 
- The number of available regular expressions and their specific capabilities can vary by database vendor, database version, and database character set.



#### Sample Use of regex

- Example,this query returns cities with district names containing [v, w, x , y, z] that are 3 million people and higher in population:

    ```SQL
    select name, district, population
    from city
    where district regexp '[vwxyz]'
    and population > 3000000
    ```

- This query uses a regular expression to list the cities with names that contain at least two occurrences of the letter *z*.

    ```SQL
    select name, district, population
    from city
    where lower(name)regexp 'z{2}';
    ```

- Next, a regular expression finds the cities with district names that contain at least one of the uppercase characters W, X, Y, or Z. This query also states that the cities must have populations greater than 3 million people.

    ```SQL
    select name, district, population
    from city
    where district regexp '[W-Z]'
    and population > 3000000
    ```

- Finally, this example query uses a regular expression to find the cities with populations that contain at least one of the digits 4, 5, or 6. Also, the cities must have populations greater than 8 million people.

    ```SQL
    select name, district, population
    from city
    where population regexp '[4-6]'
    and population > 8000000
    ```

### Working with Functions

- `SUBSTRING` functions take a substring that starts at a specific position and gathers a specified length of characters.

- A `SUBSTRING` function begins with `SUBSTRING`, and the source, starting position, and length are all contained within its parentheses. 

    ```SQL
    SELECT Name, SUBSTRING (Name, 1, 10) Name, SUBSTRING ('USA', 1,2) CountryCode 
    FROM City
    WHERE CountryCode = 'USA';
    ```

- The `CONCAT` function returns a string that is the combination of the input strings.
    ```SQL
    SELECT Code, CONCAT (Name, Continent) AS Location
    FROM country
    ```
    
- To remove the trailing spaces and insert a comma to better organize the output, the `TRIM` functions can be used within the `CONCAT` function. 
    ```SQL
    SELECT Code, CONCAT (TRIM(Name), ', ', TRIM(Continent)) AS Location
    FROM country
    ```

### Organizing Data

- Summarize the query results using `ROLLUP`
    ```SQL
    select continent, sum(population) AS 'total_pop'
    from country
    group by continent with rollup
    ```



- SQL provides a `COALESCE` function to replace this NULL value with something more meaningful in the query output.
    ```SQL
    select coalesce(continent, 'World Population') as continent,
       sum(population) AS TOTAL_POP
    from country
    group by continent with rollup
    ```

- The `ROLLUP` keyword can also be used to create summaries for subgroups within a query. This example uses `ROLLUP` to create a subtotal of the populations for each region plus an overall total for all regions in the query.
    ```SQL
    select coalesce(continent,’’),
       coalesce(region,'**Total**') as region,
       sum(population) AS total_pop
    from country
    where continent in ('Asia','Africa')
    group by continent, region with rollup
    ```

#### Ranking Data
